```python
import pandas as pd
import numpy as np
```

```python
m5_data = pd.read_csv('../Features_New/M5_features_clean.csv')
m5_data.head()
```

```python
m5_data.describe()
```

```python
columns_to_scale = ['CDRGLOB', 'GDS']
for col in columns_to_scale:
    max = m5_data[col].max()
    min = m5_data[col].min()
    m5_data[col] = (m5_data[col] - min) / (max - min)

m5_data.to_csv('../Features_New/M5_features_scaled.csv', index=False)
```
