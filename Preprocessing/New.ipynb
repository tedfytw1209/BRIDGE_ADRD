# Observation Periods

```python
def create_obs_periods_ehr(df_long, ehr_observation_path, crosswalk_path, output_folder='Features', output_filename='observation_periods_ehr.csv'):

    os.makedirs(output_folder, exist_ok=True)
    output_path = os.path.join(output_folder, output_filename)

    # Load crosswalk
    crosswalk = pd.read_csv(crosswalk_path)
    crosswalk_renamed = crosswalk.rename(columns={'nacc_person_id': 'person_id', 'SUBID': 'subid'})
    crosswalk_renamed['person_id'] = crosswalk_renamed['person_id'].astype('int64')
    print(f"Crosswalk records: {len(crosswalk_renamed):,}")

    # Load EHR observation periods
    ehr_obs = pd.read_csv(ehr_observation_path, delimiter='|')
    print(f"EHR observation_period records: {len(ehr_obs):,}")

    # Convert dates
    ehr_obs['observation_period_start_date'] = pd.to_datetime(ehr_obs['observation_period_start_date'])

    # Get earliest EHR date per person_id
    ehr_first_encounter = ehr_obs.groupby('person_id').agg({
        'observation_period_start_date': 'min', 'observation_period_end_date': 'max'
    }).reset_index()
    ehr_first_encounter.columns = ['person_id', 'ehr_first_date', 'ehr_last_date']
    print(f"Unique patients in EHR OBSERVATION: {len(ehr_first_encounter):,}")

    # Merge with crosswalk to get subid
    ehr_with_subid = pd.merge(ehr_first_encounter, crosswalk_renamed[['person_id', 'subid']], on='person_id', how='inner')
    print(f"Patients with both EHR and subid: {len(ehr_with_subid):,}")

    # Process NACC df_long for diagnosis dates
    df_long['visitdate'] = pd.to_datetime(df_long['visitdate'])

    observation_periods = []
    total_patients = 0
    successful_dementia = 0
    successful_mci = 0
    successful_normal = 0
    filtered_no_ehr = 0
    filtered_ehr_after_diagnosis = 0

    for subid, patient_data in df_long.groupby('subid'):
        total_patients += 1
        patient_data = patient_data.sort_values('visitdate')

        # Get EHR start date for this patient
        ehr_data = ehr_with_subid[ehr_with_subid['subid'] == subid]

        if len(ehr_data) == 0:
            filtered_no_ehr += 1
            continue  # Skip patient: no EHR data for this patient

        ehr_start = ehr_data['ehr_first_date'].iloc[0]
        ehr_end = ehr_data['ehr_last_date'].iloc[0]

        # Get NACC visit dates
        first_nacc_visit = patient_data['visitdate'].min()
        last_nacc_visit = patient_data['visitdate'].max()

        # Find diagnosis dates from NACC
        dementia_visits = patient_data[patient_data['DIAGNOSIS'].str.contains('Dementia', case=False, na=False)]
        mci_visits = patient_data[patient_data['DIAGNOSIS'].str.contains('MCI', case=False, na=False)]

        # Determine label and index date (from NACC diagnoses)
        if len(dementia_visits) > 0:
            label = 'Dementia'
            index_date = dementia_visits['visitdate'].min()
        elif len(mci_visits) > 0:
            label = 'MCI'
            index_date = mci_visits['visitdate'].min()
        else:
            label = 'Normal'
            index_date = last_nacc_visit  # or could use ehr_end

        # Use EHR first encounter as observation start
        obs_start_date = ehr_start

        # Observation end: 1 day before index (to prevent data leakage)
        obs_end_date = index_date - pd.Timedelta(days=1)

        # Check if EHR start is before diagnosis
        if obs_start_date >= index_date:
            filtered_ehr_after_diagnosis += 1
            continue  # Skip: EHR started after/at diagnosis

        # Calculate observation duration
        obs_duration_years = (obs_end_date - obs_start_date).days / 365.25
        time_to_index_years = (index_date - obs_start_date).days / 365.25

        observation_periods.append({
            'subid': subid, 'label': label, 'obs_start_date': obs_start_date, 'obs_end_date': obs_end_date,
            'index_date': index_date, 'ehr_first_date': ehr_start, 'ehr_last_date': ehr_end,
            'first_nacc_visit': first_nacc_visit, 'last_nacc_visit': last_nacc_visit, 'obs_duration_years': obs_duration_years,
            'time_to_index_years': time_to_index_years, 'obs_duration_days': (obs_end_date - obs_start_date).days,
            'n_nacc_visits': len(patient_data), 'first_diagnosis_date': index_date if label != 'Normal' else None
        })

        if label == 'Dementia': successful_dementia += 1
        elif label == 'MCI': successful_mci += 1
        else: successful_normal += 1

    observation_periods_df = pd.DataFrame(observation_periods)
    # Add person_id from crosswalk
    observation_periods_df = pd.merge(observation_periods_df, crosswalk_renamed[['subid', 'person_id']], on='subid', how='left')

    # Reorder columns
    cols = ['person_id', 'subid', 'label', 'obs_start_date', 'obs_end_date', 'index_date'] + \
           [col for col in observation_periods_df.columns
            if col not in ['person_id', 'subid', 'label', 'obs_start_date', 'obs_end_date', 'index_date']]
    observation_periods_df = observation_periods_df[cols]

    print("\nCOHORT SUMMARY (EHR + NACC)")
    print(f"{'='*50}")
    print(f"Total patients in df_long: {total_patients:,}")
    print(f"\nIncluded patients (>= 1 day EHR to diagnosis):")
    print(f"  Dementia: {successful_dementia:,}")
    print(f"  MCI: {successful_mci:,}")
    print(f"  Normal: {successful_normal:,}")
    print(f"  Total: {len(observation_periods_df):,}")
    print(f"\nFiltered out:")
    print(f"  No EHR data: {filtered_no_ehr:,}")
    print(f"  EHR started after diagnosis: {filtered_ehr_after_diagnosis:,}")
    print(f"  Total filtered: {total_patients - len(observation_periods_df):,}")

    if len(observation_periods_df) > 0:
        print(f"\nLabel Distribution:")
        label_counts = observation_periods_df['label'].value_counts()
        for label, count in label_counts.items():
            percentage = (count / len(observation_periods_df)) * 100
            print(f"  {label}: {count:,} ({percentage:.1f}%)")

    observation_periods_df.to_csv(output_path, index=False)
    print(f"\nObservation periods saved to: {output_path}")
    print(f"Shape: {observation_periods_df.shape}")
    print(f"Columns: {list(observation_periods_df.columns)}")

    return observation_periods_df


observation_periods_df = create_obs_periods_ehr(
    df_long, './IUSM_NACC_OMOP_20240515/OBSERVATION_PERIOD.csv','./IUSM_NACC_OMOP_20240515/NACC-omop-to-studyID-crosswalk-2024-10-15-FromEvgenia.txt',
    output_folder='Features', output_filename='observation_periods.csv'
)
```

# M1 Modality

```python
def extract_m1(df_cross, df_person, obs_periods_path, output_folder='Features', output_filename='m1_demographics.csv'):

    os.makedirs(output_folder, exist_ok=True)
    output_path = os.path.join(output_folder, output_filename)

    print(f"Cross.csv Shape: {df_cross.shape,}")
    print(f"Columns: {df_cross.columns.tolist()}")
    print(f"\nPerson Shape: {df_person.shape}")

    obs_periods = pd.read_csv(obs_periods_path)
    print(f"Our cohort: {len(obs_periods):,} patients")
    obs_periods['obs_end_date'] = pd.to_datetime(obs_periods['obs_end_date'])

    # Merge person data with observation periods for birth year
    age_data = pd.merge(obs_periods[['person_id', 'subid', 'obs_end_date']], df_person[['person_id', 'year_of_birth', 'month_of_birth']], on='person_id', how='left')
    age_data['AGE'] = age_data['obs_end_date'].dt.year - age_data['year_of_birth']      # Age at obs_end_date
    age_data['month_adjustment'] = (age_data['obs_end_date'].dt.month < age_data['month_of_birth']).astype(int)
    age_data['AGE'] = age_data['AGE'] - age_data['month_adjustment']
    print(f"Age range: {age_data['AGE'].min():.0f} - {age_data['AGE'].max():.0f} years")
    cohort_subids = obs_periods[['subid']].drop_duplicates()

    m1_features = pd.merge(cohort_subids, df_cross, on='subid', how='left')
    m1_features = pd.merge(m1_features, age_data[['subid', 'AGE']], on='subid', how='left')
    missing_demo = m1_features['SEX'].isna().sum()
    print(f"Successfully matched: {len(m1_features) - missing_demo:,} patients")
    print(f"Missing demographics: {missing_demo:,} patients")

    if missing_demo > 0:
        print(f"\n{missing_demo} patients from OMOP not found in cross.csv")
        unmapped_subids = m1_features[m1_features['SEX'].isna()]['subid'].head(10).tolist()
        print(f"Sample unmapped subids: {unmapped_subids}")

    m1_features['age<50'] = (m1_features['AGE'] < 50).astype(int)
    m1_features['age_50-65'] = ((m1_features['AGE'] >= 50) & (m1_features['AGE'] < 65)).astype(int)
    m1_features['age_65-75'] = ((m1_features['AGE'] >= 65) & (m1_features['AGE'] < 75)).astype(int)
    m1_features['age_75+'] = (m1_features['AGE'] >= 75).astype(int)

    # Map SEX: F->0, M->1
    m1_features['SEX'] = m1_features['SEX'].str[0].map({'F': 0, 'M': 1})
    for col in ['HISPANIC', 'RACE']:
        if col in m1_features.columns:
            # Isolate non-missing values to apply string operations
            valid_entries = m1_features[col].notna()
            # Extracts the part before the '=' sign
            m1_features.loc[valid_entries, col] = m1_features.loc[valid_entries, col].str.split('=', expand=True)[0].str.strip()
            m1_features[col] = pd.to_numeric(m1_features[col], errors='coerce') # Convert to numbers

    m1_columns = ['subid', 'SEX', 'HISPANIC', 'RACE', 'age<50', 'age_50-65', 'age_65-75', 'age_75+']
    missing_cols = [col for col in m1_columns if col not in m1_features.columns]
    if missing_cols:
        print(f"\nMissing columns in cross.csv: {missing_cols}")
        m1_columns = [col for col in m1_columns if col in m1_features.columns]
    m1_features = m1_features[m1_columns]

    print("\nM1 Feature Summary:")
    for col in m1_columns[1:]:
        if col in m1_features.columns:
            valid_count = m1_features[col].notna().sum()
            percentage = (valid_count / len(m1_features)) * 100
            print(f"  {col}: {valid_count:,} ({percentage:.1f}%)")
            if col in ['SEX', 'HISPANIC', 'RACE']: print(f"    Distribution: {m1_features[col].value_counts().to_dict()}")

    m1_features.to_csv(output_path, index=False)
    print(f"\nM1 features saved to: {output_path}")
    print(f"Shape: {m1_features.shape}")
    print(f"Columns: {m1_features.columns.tolist()}")

    return m1_features


m1_features = extract_m1(df_cross, df_person, 'Features/observation_periods.csv')
```

```python
m3_long_extraction = {
    'BMI': {'column': 'BMI', 'description': 'Body Mass Index'},
    'systolic_bp': {'column': 'BPSYS', 'description': 'Systolic Blood Pressure'},
    'diastolic_bp': {'column': 'BPDIAS', 'description': 'Diastolic Blood Pressure'}
}

m3_omop_extraction = {
    'total_cholesterol': {'concept_ids': [3027114, 3019900, 3049873], 'description': 'Total Cholesterol'},
    'hdl': {'concept_ids': [40757503, 3007070, 3053286, 3013473, 3023602], 'description': 'HDL Cholesterol'},
    'glucose': {'concept_ids': [3000483, 3004501, 3019210, 44816672], 'description': 'Blood Glucose'},
    'hba1c': {'concept_ids': [3034639, 3004410, 3007263, 3003309, 3005673, 40765129], 'description': 'Hemoglobin A1c'}
}

def extract_m3(df_measurement, df_long, m3_long_extraction, m3_omop_extraction, obs_periods_path, output_folder='Features', output_filename='m3_lab_tests.csv'):

    os.makedirs(output_folder, exist_ok=True)
    output_path = os.path.join(output_folder, output_filename)

    obs_periods = pd.read_csv(obs_periods_path)
    obs_periods['obs_start_date'] = pd.to_datetime(obs_periods['obs_start_date'])
    obs_periods['obs_end_date'] = pd.to_datetime(obs_periods['obs_end_date'])

    patients = obs_periods[['person_id', 'subid']].drop_duplicates()
    m3_features = patients.copy()

    # Extract from OMOP MEASUREMENT (cholesterol, HDL, glucose, HbA1c) & merge with observation periods
    df_measurement['measurement_date'] = pd.to_datetime(df_measurement['measurement_date'])
    df_measurement_temporal = pd.merge(df_measurement, obs_periods[['person_id', 'subid', 'obs_start_date', 'obs_end_date']], on='person_id', how='inner')
    print(f"Measurements for our cohort: {len(df_measurement_temporal):,}")

    # Temporal filtering
    df_measurement_temporal = df_measurement_temporal[
        (df_measurement_temporal['measurement_date'] >= df_measurement_temporal['obs_start_date']) &
        (df_measurement_temporal['measurement_date'] <= df_measurement_temporal['obs_end_date'])
    ]
    print(f"Measurements within observation windows: {len(df_measurement_temporal):,}")

    # Extract each OMOP lab test
    for lab_name, info in m3_omop_extraction.items():
        lab_data = df_measurement_temporal[df_measurement_temporal['measurement_concept_id'].isin(info['concept_ids'])]

        if len(lab_data) > 0:
            # Latest value for each patient
            latest_labs = lab_data.loc[lab_data.groupby('person_id')['measurement_date'].idxmax()][['person_id', 'value_as_number']]
            latest_labs.columns = ['person_id', lab_name]
            m3_features = pd.merge(m3_features, latest_labs, on='person_id', how='left')
            valid_count = m3_features[lab_name].notna().sum()
            percentage = (valid_count / len(patients)) * 100
            print(f"  {lab_name}: {valid_count:,} patients ({percentage:.1f}%)")
        else:
            m3_features[lab_name] = None
            print(f"  {lab_name}: No data found (0.0%)")

    df_long['visitdate'] = pd.to_datetime(df_long['visitdate'])
    # Merge with observation periods
    df_long_temporal = pd.merge(df_long, obs_periods[['subid', 'obs_start_date', 'obs_end_date']], on='subid', how='inner')
    print(f"\nLong.csv records for our cohort: {len(df_long_temporal):,}")

    # Temporal filtering
    df_long_temporal = df_long_temporal[
        (df_long_temporal['visitdate'] >= df_long_temporal['obs_start_date']) &
        (df_long_temporal['visitdate'] <= df_long_temporal['obs_end_date'])
    ]
    print(f"Long.csv records within observation windows: {len(df_long_temporal):,}")

    # Extract each long.csv feature
    for feature_name, info in m3_long_extraction.items():
        column_name = info['column']
        if column_name in df_long_temporal.columns:
            # Get non-null values
            feature_data = df_long_temporal[df_long_temporal[column_name].notna()][['subid', column_name, 'visitdate']]

            if len(feature_data) > 0:
                # Latest value for each patient
                latest_values = feature_data.loc[feature_data.groupby('subid')['visitdate'].idxmax()][['subid', column_name]]
                latest_values.columns = ['subid', feature_name]
                m3_features = pd.merge(m3_features, latest_values, on='subid', how='left')
                valid_count = m3_features[feature_name].notna().sum()
                percentage = (valid_count / len(patients)) * 100
                print(f"  {feature_name}: {valid_count:,} patients ({percentage:.1f}%)")
            else:
                m3_features[feature_name] = None
                print(f"  {feature_name}: No data found (0.0%)")
        else:
            m3_features[feature_name] = None
            print(f"  {feature_name}: Column '{column_name}' not found in long.csv (0.0%)")

    # Drop person_id, keep only subid
    m3_features = m3_features.drop('person_id', axis=1)
    m3_features.to_csv(output_path, index=False)
    print(f"\nM3 features saved to: {output_path}")
    print(f"Shape: {m3_features.shape}")
    print(f"Columns: {m3_features.columns.tolist()}")

    return m3_features


m3_features = extract_m3(df_measurement, df_long, m3_long_extraction, m3_omop_extraction, obs_periods_path='Features/observation_periods.csv', output_folder='Features', output_filename='m3_lab_tests.csv'
                         )
```

# M4 Modality

```python
target_list = [36029620, 44506794, 19092684, 925102, 1525866, 1548195, 19049038, 19049228, 1549080, 1551673, 1549254, 1525278, 1549786, 43525773, 19094980, 19029079, 19060423, 1596779, 1586808, 19041933]

def load_m4_drug_mappings(drug_files_dir='extraction_code/'):

    drug_files = {
        'NSAIDs': f'{drug_files_dir}nsaid.csv', 'Statins': f'{drug_files_dir}statins.csv',
        'Anticholinergics': f'{drug_files_dir}anticholinergics.csv', 'Hormone_Replacement': f'{drug_files_dir}hormone.csv',
        'Antihypertensives': f'{drug_files_dir}antihypertensives.csv',
        'Benzodiazepines': f'{drug_files_dir}benzodiazepines.csv', 'Proton_Pump_Inhibitors': f'{drug_files_dir}ppi.csv'
    }
    drug_mappings = {}

    for drug_class, filename in drug_files.items():
        df_drug = pd.read_csv(filename, index_col=0)
        concept_ids = df_drug['Id'].astype(int).tolist()
        drug_mappings[drug_class] = {'concept_ids': concept_ids, 'count': len(concept_ids)}
        print(f"  {drug_class}: {len(concept_ids)} concepts loaded")

    # Override Hormone_Replacement with specific target list
    target_list = [36029620, 44506794, 19092684, 925102, 1525866, 1548195, 19049038, 19049228,
                   1549080, 1551673, 1549254, 1525278, 1549786, 43525773, 19094980, 19029079,
                   19060423, 1596779, 1586808, 19041933]
    drug_mappings['Hormone_Replacement']['concept_ids'] = target_list
    drug_mappings['Hormone_Replacement']['count'] = len(target_list)
    print(f"  Hormone_Replacement: overridden with {len(target_list)} specific concepts")

    return drug_mappings


def extract_m4(df_drug_exposure, drug_mappings, obs_periods_path, output_folder='Features', output_filename='m4_medications.csv'):

    os.makedirs(output_folder, exist_ok=True)
    output_path = os.path.join(output_folder, output_filename)

    obs_periods = pd.read_csv(obs_periods_path)
    print(f"\nDrug exposure Shape: {df_drug_exposure.shape}")
    print(f"Patients with valid observation periods: {len(obs_periods):,}")

    df_drug_exposure['drug_exposure_start_date'] = pd.to_datetime(df_drug_exposure['drug_exposure_start_date'])
    obs_periods['obs_start_date'] = pd.to_datetime(obs_periods['obs_start_date'])
    obs_periods['obs_end_date'] = pd.to_datetime(obs_periods['obs_end_date'])

    df_drug_temporal = pd.merge(
        df_drug_exposure, obs_periods[['person_id', 'subid', 'obs_start_date', 'obs_end_date']], on='person_id', how='inner'
    )       # Merge drug exposures with observation periods
    print(f"Drug exposures for cohort: {len(df_drug_temporal):,}")

    df_drug_temporal = df_drug_temporal[
        (df_drug_temporal['drug_exposure_start_date'] >= df_drug_temporal['obs_start_date']) &
        (df_drug_temporal['drug_exposure_start_date'] <= df_drug_temporal['obs_end_date'])
    ]
    print(f"Drug exposures within observation windows: {len(df_drug_temporal):,}")

    # Initialize feature matrix with subid
    patients = obs_periods[['person_id', 'subid']].drop_duplicates()
    m4_features = patients.copy()
    # Extract each medication class
    for drug_class, drug_info in drug_mappings.items():
        concept_ids = drug_info['concept_ids']
        if not concept_ids:
            print(f"  {drug_class}: No concepts available")
            m4_features[drug_class] = 0
            continue
        drug_class_patients = df_drug_temporal[df_drug_temporal['drug_concept_id'].isin(concept_ids)]['person_id'].unique()

        m4_features[drug_class] = 0
        m4_features.loc[m4_features['person_id'].isin(drug_class_patients), drug_class] = 1
        prevalence = len(drug_class_patients)
        percentage = (prevalence / len(patients)) * 100
        print(f"  {drug_class}: {prevalence:,} patients ({percentage:.1f}%)")

    m4_features = m4_features.drop('person_id', axis=1)

    drug_cols = [col for col in m4_features.columns if col != 'subid']
    total_with_any_drug = m4_features[drug_cols].sum(axis=1)
    patients_with_any_drug = (total_with_any_drug > 0).sum()
    patients_with_multiple = (total_with_any_drug > 1).sum()

    print(f"\nDrug Utilization Summary:")
    print(f"  Patients with any M4 drug: {patients_with_any_drug:,} ({patients_with_any_drug/len(m4_features)*100:.1f}%)")
    print(f"  Patients with multiple drug classes: {patients_with_multiple:,} ({patients_with_multiple/len(m4_features)*100:.1f}%)")

    m4_features.to_csv(output_path, index=False)
    print(f"\nM4 features saved to: {output_path}")
    print(f"Shape: {m4_features.shape}")
    print(f"Columns: {m4_features.columns.tolist()}")

    return m4_features

drug_mappings = load_m4_drug_mappings('extraction_code/')
m4_features = extract_m4(df_drug, drug_mappings, 'Features/observation_periods.csv', )
```
