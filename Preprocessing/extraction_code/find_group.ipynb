```python
"""
mci code dictionary 
"""
mci_list = ["G31.84", "F06.7", "F06.70", "F06.71", "F09", "R41.840", "R41.841", "R41.89", "R41.9"]
mci_list_no_dots = [code.replace('.', '') for code in mci_list]
mci_list9 = ["331.83", "294.9"]
mci_list9_no_dots = [code.replace('.', '') for code in mci_list9]
mci_list = mci_list + mci_list_no_dots + mci_list9 + mci_list9_no_dots

ad_list =  ["G30", "G30.0", "G30.1", "G30.8", "G30.9"]
ad_list_no_dots = [code.replace('.', '') for code in ad_list]
ad_list9 = ['331.0']
ad_list9_no_dots = [code.replace('.', '') for code in ad_list9]
ad_list = ad_list + ad_list_no_dots + ad_list9 + ad_list9_no_dots

not_control_list = [
    "290.0", "290.10", "290.11", "290.12", "290.13", "290.2", "290.21", "290.3", "290.4", "290.40",
    "290.41", "290.42", "290.43", "290.8", "290.9", "291.2", "292.82", "294.0", "294.1", "294.11",
    "294.2", "294.21", "294.8", "331", "331.1", "331.11", "331.19", "331.2", "331.3", "331.4",
    "331.5", "331.6", "331.7", "331.81", "331.82", "331.89", "332", "332.0", "332.1", "348.81",
    "437.0", "437.1", "437.2", "437.3", "437.4", "437.5", "437.6", "437.7", "437.8", "437.9",
    "797", "A81.0", "F00", "F01", "F01.5", "F01.50", "F01.51", "F02", "F02.80", "F02.81",
    "F03", "F03.90", "F03.91", "F04", "F05", "F05.1", "F06.0", "F06.1", "F06.8", "F10.27",
    "F10.6", "F10.97", "F13.27", "F13.97", "F18.17", "F18.27", "F18.97", "F19.17", "F19.27", "F19.97",
    "G13.8", "G20", "G21", "G21.1", "G21.11", "G21.19", "G21.2", "G21.3", "G21.4", "G21.8",
    "G21.9", "G30", "G30.0", "G30.1", "G30.8", "G30.9", "G31.0", "G31.01", "G31.09", "G31.1",
    "G31.2", "G31.8", "G31.83", "G31.85", "G31.89", "G31.9", "G45.4", "G91.0", "G91.1", "G91.2",
    "G93.7", "G93.81", "G94", "I67.1", "I67.2", "I67.3", "I67.4", "I67.5", "I67.6", "I67.7",
    "I67.81", "I67.82", "I67.89", "I67.9", "R41.81", "R54", "294.9", "331.83", "G31.84", "F06.7",
    "F06.71", "F06.70", "F09"
]
not_control_list_no_dots = [code.replace('.', '') for code in not_control_list]

not_control_list = not_control_list + not_control_list_no_dots + mci_list + ad_list
```

```python
def get_outcome(row):
#     extract each row's outcome for each subject
        value = row.condition_source_value
        mci = 0
        ad = 0
        control = 0
        if len(set(value).intersection(set(mci_list))) != 0:
            mci = 1
        if len(set(value).intersection(set(ad_list))) != 0:
            ad = 1
        if len(set(value).intersection(set(not_control_list))) == 0:
            control = 1
        row['outcome_mci'] = mci
        row['outcome_ad'] = ad
        row['outcome_control'] = control
        return row
    
def return_outcome(df):
    df = df.apply(get_outcome, axis = 1)
    df.condition_start_date = pd.to_datetime(df.condition_start_date)
    df.birth_datetime = pd.to_datetime(df.birth_datetime)
    df = df.sort_values(by ='condition_start_date', ascending = True)
#     each patient time period
    df['time_period'] = np.floor((df.iloc[-1,1] - df.iloc[0,1]).days/365.25)
#     adding index age and oct age
    
#     control
    if (df.outcome_mci.sum() == 0) and (df.outcome_ad.sum() == 0) and (df.outcome_control.sum() == 0):
        df['outcome_sum'] = 'not case/control'
        df['index_age'] = np.floor((pd.to_datetime(df.iloc[0,1]) - pd.to_datetime(df.iloc[0,2])).days / 365.25)
        df['first_diag_date'] = df.iloc[0,1]
        df['first_diag_date(ad)'] = np.nan
        return df.head(1)
    elif (df.outcome_mci.sum() == 0) and (df.outcome_ad.sum() == 0) and (df.outcome_control.sum() != 0):
        df['outcome_sum'] = 'control'
        df['index_age'] = np.floor((pd.to_datetime(df.iloc[0,1]) - pd.to_datetime(df.iloc[0,2])).days / 365.25)
        df['first_diag_date'] = df.iloc[0,1]
        df['first_diag_date(ad)'] = np.nan
        df.outcome_control = 1
        return df.head(1)
#         mci
    elif (df.outcome_mci.sum() != 0) and (df.outcome_ad.sum() == 0):
        df['outcome_sum'] = 'mci'
#         first mci diagnosis index age
        df['index_age'] = np.floor((pd.to_datetime(df[df.outcome_mci == 1].head(1).condition_start_date.values[0]) - pd.to_datetime(df[df.outcome_mci == 1].head(1).birth_datetime.values[0])).days / 365.25)
        df_start = df[df.outcome_mci == 1]
        df_start['first_diag_date'] = df_start.iloc[0,1]
        df_start['first_diag_date(ad)'] = np.nan
        return df_start.head(1)
#     ad
    elif (df.outcome_mci.sum() == 0) and (df.outcome_ad.sum() != 0):
        df['outcome_sum'] = 'ad'
#         first ad diagnosis index age
        df['index_age'] = np.floor((pd.to_datetime(df[df.outcome_ad == 1].head(1).condition_start_date.values[0]) - pd.to_datetime(df[df.outcome_ad == 1].head(1).birth_datetime.values[0])).days / 365.25)
        df_start = df[df.outcome_ad == 1]
        df_start['first_diag_date'] = df_start.iloc[0,1]
        df_start['first_diag_date(ad)'] = np.nan
        return df_start.head(1)
#     ad + mci
    elif (df.outcome_mci.sum() != 0) and (df.outcome_ad.sum() != 0):
#         for the ad_mci, their condition start date is first mci date, index age is ad age
        df['outcome_sum'] = 'mci + ad'
        df['index_age'] = np.floor((pd.to_datetime(df[df.outcome_ad == 1].head(1).condition_start_date.values[0]) - pd.to_datetime(df[df.outcome_ad == 1].head(1).birth_datetime.values[0])).days / 365.25)
        df_ad_date = df[df.outcome_ad == 1].iloc[0,1]
        df_start = df[df.outcome_mci == 1]
        df_start['first_diag_date'] = df_start.iloc[0,1]
        df_start['first_diag_date(ad)'] = df_ad_date
        df_start.outcome_ad = 1
        return df_start.head(1)
    else:
        print('unknown')
        return None
```

```python
# cannot directly read df_agg since the list format will change
from tqdm import tqdm 
tqdm.pandas()
result = df_agg.reset_index(drop=True).groupby('person_id').progress_apply(return_outcome)
result.reset_index(drop = True, inplace = True)
```

```python
result.to_csv('outcome.csv', index=False)
```
