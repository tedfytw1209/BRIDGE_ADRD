```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
# control
control_cci = pd.read_csv('control_cci.csv')
demog = pd.read_csv('control_demog.csv')
demog['birth_year'] = pd.to_datetime(demog.birth_datetime).dt.year
control = pd.merge(demog, control_cci, on='person_id', how= 'inner')[['person_id', 'birth_year', 'gender_source_value', 'race_ethnicity', 'cci_score']]
control = control[control.gender_source_value != 'UNKNOWN']
gender_encoded = pd.get_dummies(control["gender_source_value"], prefix="gender")
race_encoded = pd.get_dummies(control["race_ethnicity"], prefix="race")
gender_encoded = gender_encoded.astype(int)
race_encoded = race_encoded.astype(int)

control = pd.concat([control, gender_encoded, race_encoded], axis=1)


control.drop(columns=["gender_source_value", "race_ethnicity"], inplace=True)
control['ad'] = 0

for year in ['1year', '3year', '5year']:
    # case
    cci1 = pd.read_csv(f'case_cci_{year}.csv')
    demog = pd.read_csv('case_demog.csv')
    demog['birth_year'] = pd.to_datetime(demog.birth_datetime).dt.year
    keep = pd.merge(demog, cci1, on='person_id', how= 'inner')[['person_id','birth_year', 'gender_source_value', 'race_ethnicity', 'cci_score']]
    gender_encoded = pd.get_dummies(keep["gender_source_value"], prefix="gender")
    race_encoded = pd.get_dummies(keep["race_ethnicity"], prefix="race")
    gender_encoded = gender_encoded.astype(int)
    race_encoded = race_encoded.astype(int)


    df_encoded = pd.concat([keep, gender_encoded, race_encoded], axis=1)

    df_encoded.drop(columns=["gender_source_value", "race_ethnicity"], inplace=True)
    df_encoded['ad'] = 1


    X = pd.concat([control, df_encoded])

    model = LogisticRegression()

    model.fit(X.iloc[:,1:-1], X.iloc[:,-1])

    X1 = X.copy()
    X1['score'] = model.predict_proba(X.iloc[:,1:-1])[:, 1]
    case = X1[X1.ad == 1]
    control1 = X1[X1.ad == 0]
    print('case and control', case.shape, control1.shape)

    
    
    
    all_match = pd.DataFrame()
    match_times = len(control1) // len(case)
    print(control1.shape)
    for i in range(match_times):
        case1 = case.copy()
        while not case1.empty and not control1.empty:
            nn = NearestNeighbors(n_neighbors=1)
            nn.fit(control1[["score"]])  

            distances, indices = nn.kneighbors(case1[["score"]])  
            matched_index = indices.flatten()[0] 


            matched_case = case1.iloc[0]
            matched_control = control1.iloc[matched_index]


            matched_pair = pd.DataFrame({
                "case_id": [matched_case["person_id"]],
                "case_score": [matched_case["score"]],
                "control_id": [matched_control["person_id"]],
                "control_score": [matched_control["score"]],
            })
            all_match = pd.concat([all_match, matched_pair], ignore_index=True)


            control1 = control1.drop(index=matched_index).reset_index(drop=True)


            case1 = case1.iloc[1:].reset_index(drop=True)


    print(all_match.shape)
    print('case uniuqe', all_match.case_id.nunique())
    print('control uniuqe', all_match.control_id.nunique())
    all_match.to_csv(f'match_{year}.csv', index=False)
```

```python
import re
def get_m5(row):
    text = str(row.text)
    # moca
    moca_list = [
        r"moca total score[ ]*total score .*?:.*?(\d{1,2})",
        r"(\d{1,2}/30)(?!/).*on the .*moca",
        r"(?:moca|montreal cognitive assessment) = ?(\d{1,2})(?!/)",
        r"(?:moca|montreal cognitive assessment) (?!education).*?(\d{1,2}/30)(?!/)",
        r"(?:moca|montreal cognitive assessment) (?!education).*?(\d{1,2} out of 30)(?!/)",
        r"(?:moca|montreal cognitive assessment):(?!education).*?(\d{1,2}/30)(?!/)",
        r"MOCA\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){1}(?!\d).*?orientation.*?total score.*?(?:out of 30)\) ?((\d{1,2} ){1})",
        r"MOCA\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){2}(?!\d).*?orientation.*?total score.*?(?:out of 30)\) ?((\d{1,2} ){2})",
        r"MOCA\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){3}(?!\d).*?orientation.*?total score.*?(?:out of 30)\) ?((\d{1,2} ){3})",
        r"MOCA\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){4}(?!\d).*?orientation.*?total score.*?(?:out of 30)\) ?((\d{1,2} ){4})",
        r"MOCA\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){5}(?!\d).*?orientation.*?total score.*?(?:out of 30)\) ?((\d{1,2} ){5})"
    ]

    # The `val` specifying the group to extract
    val = "group(1)"

    # Apply each regex and extract the score
    moca_values = []
    for expression in moca_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Dynamically extract group(1)
            moca_values.append(score)
    row['moca'] = moca_values

    # Define the list of regex patterns for MMSE
    mmse_list = [
        r"(?:mmse|mini-mental status exam) (?!=).*?(\d{1,2}/30)",
        r"(?:mmse|mini-mental status exam) ?((\d{1,2}/30 ){1,})",
        r"(?:mmse|mini-mental status exam): .*?((\d{1,2}/30 ){1,})",
        r"mmse = ?(\d{2})(?!/)",
        r"(?:mmse|mini-mental status exam) = ?(\d{1,2}/30)",
        r"mini mental status exam .* ?(\d{1,2}/30)",
        r"(\d{1,2}/30) on mmse",
        r"mmse\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){1}(?!\d).*?orientation.*?total score.*?((\d{1,2} ){1})",
        r"mmse\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){2}(?!\d).*?orientation.*?total score.*?((\d{1,2} ){2})",
        r"mmse\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){3}(?!\d).*?orientation.*?total score.*?((\d{1,2} ){3})",
        r"mmse\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){4}(?!\d).*?orientation.*?total score.*?((\d{1,2} ){4})",
        r"mmse\) (?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2} ){5}(?!\d).*?orientation.*?total score.*?((\d{1,2} ){5})",
        r"mmse\) (?![ ]{0,}folstein)(?!([0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2})).*?total score.*?(\d{1,2})",
        r"mmse: ?(\d{1,2} )",
        r"mini mental state exam: ?(\d{1,2})",
        r"total mmse ?(\d{1,2})",
        r"(?:mini-mental status|mini-mental state examination) .*?(\d{1,2}) out of 30"
    ]


    # List to store all matched values
    mmse_values = []

    # Apply each regex and extract the score
    for expression in mmse_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Extract group(1) dynamically
            mmse_values.append(score)
            
    row['mmse'] = mmse_values
    # apoe
    apoe_list = [
        r"(?:apoe|apolipoprotein e).*?(\d{1}/\d{1})",
        r"(\d{1}/\d{1}) apoe genotype"
    ]


    # List to store all matched values
    apoe_values = []

    # Apply each regex and extract the value
    for expression in apoe_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Extract group(1) dynamically
            apoe_values.append(score)
            
    row['apoe'] = apoe_values
    # drs
    drs_list = [
        r"overall cdr ?([0-9]\.[0-9]|[0-9])",
        r"clinical dementia rating scale category of[ ]{1,}?([0-9]\.[0-9]|[0-9])",
        r"clinical dementia rating scale\):([0-9])",
        r"cdr total = ([0-9])",
        r"clinical dementia rating scale.*sum of boxes = ([0-9])",
        r"clinical dementia rating scale.*overall cdr = ([0-9])",
        r"clinical dementia rating scale is ([0-9])"
    ]

    # List to store all matched values
    drs_values = []

    # Apply each regex and extract the value
    for expression in drs_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Extract group(1) dynamically
            drs_values.append(score)
            
    row['drs'] = drs_values
    # cog
    cog_list = [
        r"mini cog.*?(?:[0-3]?[0-9]/[0-3]?[0-9]/(?:[0-9]{2})?[0-9]{2}).*negative screen ?(\d{1})",
        r"mini cog:[ ]{0,}?(\d{1}/5|\d{1} )",
        r"(?:mini-cog|mini cog score)[ ]{1,}?(\d{1}/5|\d{1} )"
    ]

    # List to store all matched values
    cog_values = []

    # Apply each regex and extract the value
    for expression in cog_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Extract group(1) dynamically
            cog_values.append(score)

    row['cog'] = cog_values
    # Define the list of regex patterns for GDS
    gds_list = [
        r"gds-(\d{1,2})",
        r"geriatric depression scale score (?:.*): ?(\d{1,2})",
        r"(?:geriatric depression scale|gds)[ ]{1,}(?:= ){0,}?(\d{1,2}/[ ]{0,}\d{1,2}|\d{1,2})",
        r"(?:geriatric depression scale|gds),[ ]{1,}?(\d{1,2}/\d{1,2})",
        r"(?:geriatric depression scale|gds) ?(\d{1,2})"
    ]


    # List to store all matched values
    gds_values = []

    # Apply each regex and extract the value
    for expression in gds_list:
        match = re.search(expression, text, re.IGNORECASE)
        if match:
            score = eval(f"match.{val}")  # Extract group(1) dynamically
            gds_values.append(score)
    row['gds'] = gds_values
    return row
```

```python
def process_row(df):
    columns_to_check = ['moca', 'mmse', 'apoe', 'drs', 'cog', 'gds']
    processed = {"person_id": df["person_id"].iloc[-1]}  # Keep the last person_id
    for col in columns_to_check:
        if col == "apoe":  # Special case for 'apoe'
            # Record 1 if there is any value in the column, otherwise 0
            processed[col] = 1 if df[col].apply(lambda x: isinstance(x, list) and len(x) > 0).any() else 0
        else:
            # For other columns, get the last non-empty list
            non_empty_values = df[col].apply(lambda x: x if isinstance(x, list) and len(x) > 0 else np.nan)
            if non_empty_values.notna().any():
                processed[col] = non_empty_values[non_empty_values.notna()].iloc[-1]
            else:
                processed[col] = np.nan
    return processed
```

```python
def get_m4(row):
    all_list = [str(i) for i in row.drug_concept_id]  
    if len(set(all_list).intersection(set(NSAIDs))) != 0:
        row['NSAIDs'] = 1
    else:
        row['NSAIDs'] = 0

    if len(set(all_list).intersection(set(statins))) != 0:
        row['statins'] = 1
    else:
        row['statins'] = 0

    if len(set(all_list).intersection(set(anticholinergics))) != 0:
        row['anticholinergics'] = 1
    else:
        row['anticholinergics'] = 0

    if len(set(all_list).intersection(set(hormones))) != 0:
        row['hormones'] = 1
    else:
        row['hormones'] = 0

    if len(set(all_list).intersection(set(antihypertensives))) != 0:
        row['antihypertensives'] = 1
    else:
        row['antihypertensives'] = 0

    if len(set(all_list).intersection(set(benzodiazepines))) != 0:
        row['benzodiazepines'] = 1
    else:
        row['benzodiazepines'] = 0

    if len(set(all_list).intersection(set(PPI))) != 0:
        row['PPI'] = 1
    else:
        row['PPI'] = 0
    
    return row
```

```python
def get_bmi(row):
    if pd.isna(row.weight) or pd.isna(row.height):
        row.bmi = np.nan
    else:
        row.bmi = round(row.weight / ((row.height / 100) ** 2),2)
    return row
```
